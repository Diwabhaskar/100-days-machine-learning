### opening a csv file from an url

```python
import requests
from io import StringIO

url = " "  # yaha par kisi url ka paste karenge tab jake chalega
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers= headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### Skip bad lines

### dtypes parameter

```python
df = pd.read_csv('academic Stress level - maintainance 1.csv', dtype={"Age": int})
df
```

### Convertors

```python
def rename(name):
    if name == "undergraduate":
        return "UG"
    else :
        return name
```

```python
rename("undergraduate")
```

```python
df = pd.read_csv('academic Stress level - maintainance 1.csv',converters={'Your Academic Stage':rename})
df
```

### na_values parameter

```python
df = pd.read_csv('academic Stress level - maintainance 1.csv', na_values=['No'])
df
```

### loading a huge dataset in chunks

```python
df = pd.read_csv('academic Stress level - maintainance 1.csv',chunksize= 30)
df
```
